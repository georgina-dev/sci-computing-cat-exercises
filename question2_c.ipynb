```python
import numpy as np
```

```python
# logistic rgression coefficients
b0 = -1
b1 = 0.4
```

```python
# function to calculate probability
def probability(x):
  log_odds = b0 + b1 * x
  p = 1 / (1 + np.exp(-log_odds))
```

```python
# calculate probabilities
p0 = probability(0)
p1 = probability(1)

print("Probability at 0 visits: ", p0)
print("Probability at 1 visits: ", p1)
```

```python
# calculate odds ratio
odds_ratio = np.exp(b1)

print("Odds Ratio: ", odds_ratio)
```

```python
# percentage increase in odds
percentage_increase = (odds_ratio - 1) * 100
print("Percentage increase in odds:",
      percentage_increase, "%")
```
