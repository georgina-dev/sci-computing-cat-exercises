```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# original waiting times in minutes
waiting_times = np.array([8, 10, 12, 15, 18, 20, 25, 40])
```

```python
# set fixed random seed
np.random.seed(42)
```

```python
# number of bootstrap means
B = 1000
```

```python
# store bootstrap means
bootstrap_means = []
```

```python
# perfomr bootstrap resampling
for i in range(B):
  boostrap_sample = np.random.choice(
      waiting_times,
      size=len(waiting_times),
      replace=True
  )
  bootstrap_means.append(np.mean(boostrap_sample))
```

```python
# convert to array
bootstrap_means = np.array(bootstrap_means)
```

```python
# original sample mean
original_mean = np.mean(waiting_times)
```

```python
# bootstrap standard error
bootstrap_se = np.std(bootstrap_means, ddof=1)
```

```python
# 95% percentile confidence interval
lower = np.percentile(bootstrap_means, 2.5)
upper = np.percentile(bootstrap_means, 97.5)
```

```python
# display results
print("Original Mean: ", original_mean)
print("Boostrap Standard Error: ", bootstrap_se)
print("95% confidence Interval: ", (lower, upper))
```

```python
# plot bootstrap distribution
plt.hist(bootstrap_means, bins=25, edgecolor='black')
plt.xlabel("Boostrap Means (minutes)")
plt.ylabel("Frequency")
plt.title("Bootstrap Distribution of Mean waiting Times")
plt.show()
```
